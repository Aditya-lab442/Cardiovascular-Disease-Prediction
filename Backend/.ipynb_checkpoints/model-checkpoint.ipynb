```python
# Task 5 - Model Evaluation, Cross-Validation, Comparison and Hyperparameter Tuning
# Cardiovascular Disease Classification

import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

RANDOM_STATE = 42
```

```python
# 1. Load and prepare the same dataset used by the existing project
df = pd.read_csv("cardio_train.csv")

X = df.drop(columns=["cardio", "id"]).copy()
y = df["cardio"]

# The original dataset stores age in days. The API accepts age in years,
# so the training feature is also converted to years.
X["age"] = X["age"] / 365.25

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

print("Dataset shape:", df.shape)
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
# 2. Define the classification models required by Task 5
models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(max_iter=1000, random_state=RANDOM_STATE))
    ]),
    "Random Forest": Pipeline([
        ("scaler", StandardScaler()),
        ("model", RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE, n_jobs=-1))
    ]),
    "AdaBoost": Pipeline([
        ("scaler", StandardScaler()),
        ("model", AdaBoostClassifier(n_estimators=100, random_state=RANDOM_STATE))
    ]),
    "Gradient Boosting": Pipeline([
        ("scaler", StandardScaler()),
        ("model", GradientBoostingClassifier(random_state=RANDOM_STATE))
    ])
}
```

```python
# 3. Evaluation + overfitting/underfitting check + 5-fold cross-validation
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
results = []

for name, model in models.items():
    model.fit(X_train, y_train)

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_score = accuracy_score(y_train, train_pred)
    test_score = accuracy_score(y_test, test_pred)

    fold_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring="accuracy", n_jobs=1)

    results.append({
        "Model": name,
        "Train Accuracy": train_score,
        "Test Accuracy": test_score,
        "Precision": precision_score(y_test, test_pred, zero_division=0),
        "Recall": recall_score(y_test, test_pred, zero_division=0),
        "F1-score": f1_score(y_test, test_pred, zero_division=0),
        "CV Mean": fold_scores.mean(),
        "CV Std": fold_scores.std(),
        "CV Min": fold_scores.min(),
        "CV Max": fold_scores.max()
    })

results_df = pd.DataFrame(results).sort_values(
    by=["CV Mean", "CV Std"], ascending=[False, True]
).reset_index(drop=True)

print("Classification Model Comparison")
display(results_df.round(4))
```

```python
# 4. Interpret the train/test scores

print("\nFit check:")

for _, row in results_df.iterrows():
    gap = row["Train Accuracy"] - row["Test Accuracy"]

    if gap > 0.05:
        status = "Possible overfitting"

    elif row["Train Accuracy"] < 0.65 and row["Test Accuracy"] < 0.65:
        status = "Possible underfitting"

    else:
        status = "Train/Test scores are reasonably close"

    print(
        f'{row["Model"]}: '
        f'train={row["Train Accuracy"]:.4f}, '
        f'test={row["Test Accuracy"]:.4f}, '
        f'gap={gap:.4f} -> {status}'
    )


print("\nCross-validation interpretation:")

print(
    "CV Mean = average score across 5 folds; "
    "CV Std = score spread. "
    "A larger CV Std indicates more variation between folds."
)
```

```python
# 5. Select the model using the highest 5-fold CV mean; CV Std is used as the tie-breaker.
best_model_name = results_df.iloc[0]["Model"]
best_cv_mean = results_df.iloc[0]["CV Mean"]
best_cv_std = results_df.iloc[0]["CV Std"]
best_model = models[best_model_name]

print("Model selected for hyperparameter tuning:", best_model_name)
print(f"CV Mean: {best_cv_mean:.4f}")
print(f"CV Std:  {best_cv_std:.4f}")
```

```python
# 6. Hyperparameter tuning with RandomizedSearchCV
# The search space is chosen according to whichever model won the CV comparison.
param_distributions = {
    "Logistic Regression": {
        "model__C": [0.01, 0.1, 0.5, 1, 2, 5, 10],
        "model__solver": ["lbfgs", "liblinear"],
        "model__class_weight": [None, "balanced"]
    },
    "Random Forest": {
        "model__n_estimators": [100, 200, 300],
        "model__max_depth": [None, 5, 10, 15, 20],
        "model__min_samples_split": [2, 5, 10],
        "model__min_samples_leaf": [1, 2, 4],
        "model__max_features": ["sqrt", "log2", None]
    },
    "AdaBoost": {
        "model__n_estimators": [50, 100, 150, 200],
        "model__learning_rate": [0.01, 0.05, 0.1, 0.5, 1.0, 1.5]
    },
    "Gradient Boosting": {
        "model__n_estimators": [50, 100, 150, 200],
        "model__learning_rate": [0.01, 0.05, 0.1, 0.15, 0.2],
        "model__max_depth": [2, 3, 4, 5],
        "model__min_samples_split": [2, 5, 10],
        "model__min_samples_leaf": [1, 2, 4],
        "model__subsample": [0.8, 1.0]
    }
}

search = RandomizedSearchCV(
    estimator=best_model,
    param_distributions=param_distributions[best_model_name],
    n_iter=10,
    scoring="accuracy",
    cv=cv,
    random_state=RANDOM_STATE,
    n_jobs=1,
    verbose=1
)

search.fit(X_train, y_train)

tuned_model = search.best_estimator_
print("Best parameters:")
print(search.best_params_)
print(f"Best CV score: {search.best_score_:.4f}")
```

```python
# 7. Re-test the tuned model on the untouched test set
tuned_train_pred = tuned_model.predict(X_train)
tuned_test_pred = tuned_model.predict(X_test)

tuned_metrics = {
    "Train Accuracy": accuracy_score(y_train, tuned_train_pred),
    "Test Accuracy": accuracy_score(y_test, tuned_test_pred),
    "Precision": precision_score(y_test, tuned_test_pred, zero_division=0),
    "Recall": recall_score(y_test, tuned_test_pred, zero_division=0),
    "F1-score": f1_score(y_test, tuned_test_pred, zero_division=0)
}

print(pd.Series(tuned_metrics).round(4))

original_test = results_df.loc[results_df["Model"] == best_model_name, "Test Accuracy"].iloc[0]
print(f"Original {best_model_name} test accuracy: {original_test:.4f}")
print(f"Tuned {best_model_name} test accuracy:    {tuned_metrics['Test Accuracy']:.4f}")
print("Test-set improvement:", tuned_metrics["Test Accuracy"] - original_test)
```

```python
# 8. Final classification comparison table required by Task 5
comparison_table = results_df[["Model", "Accuracy" if "Accuracy" in results_df.columns else "Test Accuracy", "Precision", "Recall", "F1-score"]].copy()
comparison_table = comparison_table.rename(columns={"Test Accuracy": "Accuracy"})
display(comparison_table.round(4))
```

```python
# 9. Save the tuned pipeline for the FastAPI backend
# Saving the whole pipeline means the API no longer needs a separate scaler.pkl.
joblib.dump(tuned_model, "model.pkl")

# Keep a copy of the final CV/comparison results for reference.
comparison_table.to_csv("model_comparison.csv", index=False)

print("Saved tuned model to model.pkl")
print("Saved comparison table to model_comparison.csv")
```
